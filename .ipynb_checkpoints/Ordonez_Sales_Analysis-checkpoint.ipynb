# Q2 : 
### What is monthly total revenue for in-store sales in each of the two sales territories, over the full period covered by the data?
#

```python
storesales.head() #Reviewing first  rows of dataset and columns
```

```python
sales_location = storesales.merge(storedetails, on ="Store ID") #Merging the sales dataset with the store details using store ID as common key
```

```python
ny_ct_sales = sales_location[ #Filtering the merged data to include only NY & CT
    sales_location["State"].isin(["New York", "Connecticut"]) #.isin() checks if values match NY or CT
].copy() # .copy() creates new DataFrame to avoid modifications or original data
```

```python
ny_ct_sales.columns #Displays column names confirms merge
```

```python
ny_ct_sales["Transaction Date"] = pd.to_datetime(ny_ct_sales["Transaction Date"]) #Converting Transaction Date to DateTime Format
#Allows time-based operations to be able to group by month
```

```python
ny_ct_sales["Month"] = ny_ct_sales["Transaction Date"].dt.to_period("M") #Creating new "Month" column by gathering from Transaction Date
# "M" will group into monthly periods (ex: 02-2022)
```

```python
monthly_revenue = ny_ct_sales.groupby(
    ["State", "Month"] #Grouping by State then Month
)["Sale Amount"].sum()  #Calculating the total sales revenue by sum of Sale Amount Column
```

```python
monthly_revenue # Displaying grouped monthly revenue results
```

```python
monthly_revenue = monthly_revenue.reset_index() #Data clean up using Reset index to now show State and Month as regular columns
monthly_revenue #Display DataFrame after reset index
```

```python
#Create a pivot table to reshape the data

pivot_table = monthly_revenue.pivot( #Creating pivot table for Data
    index="Month", #Set month as the rows
    columns="State", #Create columns for each state
    values="Sale Amount" #Fills the Value with total revenue 
)
```

```python
#Creating bar chart from pivot table 

sns.set_style("whitegrid")  # Apply styling before creating plots
pivot_table.plot(kind="bar", figsize=(15,6),color =["Blue","Orange"])  #Kind=bar specifies chart type/ FigSize specifies size of chart/ Color assigns specific color 
plt.title("Monthly revenue by Territory (New York VS Conneticut)") #Adds title to chart for context
plt.xlabel("Month") # Labels X-axis
plt.ylabel("Total Revenue") #Labels Y-axis
plt.xticks(rotation=45) #Rotates X-axis to avoid labels overlapping 

plt.show() #Display Chart
```

#
# Q2 OVERVIEW :

### The bar chart above compares the monthly revenue between New York and Conneticut. 
### The blue bar represents Conneticut while the Green represents New York. 
### Each bar reflects the month and amount of revenue from January 2022 - December 2025.
### The bar chart helps us identify trends and visualize which territory is performing better and the difference between revenue.

# Q3: 
### How would you rank the sales performance of each store in each sales territory? Which are the top-performing stores?
#

```python
store_performance = ny_ct_sales.groupby( #Grouping data by State and Store ID
    ["State", "Store ID"]
)["Sale Amount"].sum().reset_index() #.Sum adds all the sales for each store
#Reset index will group back to clean DataFrame

#Creating "Rank" to rank each store based on total sales .groupby makes sure each ranking is per state
store_performance["Rank"] = store_performance.groupby("State")["Sale Amount"] \
.rank(method="dense", ascending=False) # 'dense' ranking means no gaps in ranking numbers
```

```python
store_performance = store_performance.sort_values( #Using .sort_values to sort by State and Rank
    ["State", "Rank"]
) #Stores will be listed by highest to lowest performance
```

```python
top_stores = store_performance[store_performance["Rank"] <= 5] #Filters the data to only include top 5 in each state
top_stores #Displays Top-Performing stores shows State, Store ID, Total Sales, and Rank
```

```python
top_1 = store_performance[store_performance["Rank"] == 1] #Gathering best store per state

plt.figure(figsize=(6,4)) #Creating a new figure/ Figsize is setting chart size

#Creating a bar chart to compare Top-Performing store locations

sns.set_style("whitegrid") #Grid Styling 
sns.barplot( #Selecting Bar Chart
    data=top_1, #Using ONLY top-ranking stores per state 
    x="State", #Set X-axis to State
    y="Sale Amount" #Seting Y-axis to total sales for each store
)

plt.title("Top-Performing Store in Each State") #Adding title
plt.xlabel("State") #Labeling X-axis
plt.ylabel("Total Sales") #Labeling Y-axis

plt.tight_layout() #Layout to fix spacing issues
plt.show() #Displaying Chart
```

#
# Q3 OVERVIEW :
### My top performing store in Connecticut is the New London location (Store ID 870) with a total sales amount of approximately $331,116.

### For New York my top performing store is the New York location (Store ID 850) with a significantly higher sales amount at approximately $621,565. 
### While both stores provide a strong contribution to their assigned territories, 
### the New York location stands out as the highest performing store across my assigned territories with around 53% more sales.

# Q4: 
### Comparing the customer ID from the customer list data with the rewards ID from the sales data, who were the top customers in each sales territory?
#

```python
print(customerlist.columns) #Shows all columns in CustomerList
print(ny_ct_sales.columns) #Shows all columns in NY/CT dataset

#Review prior to merging info
```

```python
customerlist = pd.read_csv("customer_list.csv", sep="|") #Loading customerlist into DataFrame ALSO using sep="|" to seperate columns
```

```python
merged_data = ny_ct_sales.merge( #Merging Sales data with Customerlist
    customerlist,
    left_on="RewardsID", #Using Reward ID from Sales Data
    right_on="cust_id", #Using CustID from customer list
    how="inner" #"Inner" will keep rows where ID matches in both sets of data
)
```

```python
customer_performance = merged_data.groupby( #Group data by State, Customer ID, and name
    ["State", "cust_id","name"]
)["Sale Amount"].sum().reset_index() #Sum of Sales Amount to get total spending per customer

#Ranking customers using group by to make sure its by state
customer_performance["Rank"] = customer_performance.groupby("State")["Sale Amount"] \
    .rank(method="dense", ascending=False) #"Dense" to have no gaps in data and rank descending to show highest as Rank 1

top_customers = customer_performance[customer_performance["Rank"] <= 5] #showing top 5 customers in each territory
#Rank <=5 select highest spending customer

top_customers = top_customers.sort_values(["State", "Rank"])  # Sort by state and rank

top_customers #Displaying results
```

#
# Q4 OVERVIEW :
## The top 5 customers were based on total spending.

## Top customer in CT is Marie S. and Top customer in NY is Julian B.

# Q5: 
## What is the number of transactions per month by product category in each assigned territory?
## What is total sales revenue per month by category? What might this tell you about the most popular products, and where could there be opportunity for growth?
#

```python
ny_ct_sales['Month'] = pd.to_datetime(ny_ct_sales['Transaction Date']).dt.to_period('M') #Creating Month column from transaction date
#Converting transaction date to datetime format
```

```python
print(ny_ct_sales.columns) #Reviewing data before merging 
```

```python
ny_ct_sales = ny_ct_sales.merge( #Merging NYCTDataset with Products
    products[['Prod Num','CategoryID']], #Joined using product num as key
    on='Prod Num') 
print(ny_ct_sales.columns) #Confirming Category ID was added
```

```python
ny_ct_sales = ny_ct_sales.merge( #Mergining NYCTData set with ProductCategories
    productcategories[['CategoryID','Category']], #Joined using category ID as key
    on='CategoryID')
print(ny_ct_sales.columns) #Confirming Category was added
```

```python
#Gathering Transaction Count

transactions = ( #grouping data to count number of transactions
    ny_ct_sales
    .groupby(['Month', 'State', 'Category']) #Group by Month,State,Category
    .size() #Counts how many transactions exist in each group
    .reset_index(name='Transaction_Count') #Converts result into clean dataframe with column name Transaction Count
)
transactions.head(12) #Displaying Results first 12 rows
```

```python
#Gathering Revenue Data

revenue = ( #New data frame for revenue
    ny_ct_sales
    .groupby(['Month', 'State', 'Category'])['Sale Amount'] #Group by month state and category
    .sum() #ADding total sales for each group
    .reset_index(name='Total_Revenue') #Creating clean table including Total Revenue column
)
revenue.head(12) #Displaying Results first 12 rows
```

```python
# Merging Transaction Count and Revenue Totals
final = pd.merge( #Merging Transactions and Revenue Dataframes
    transactions,
    revenue,
    on=['Month', 'State', 'Category'], #Joining with matching rows
    how='left' #Keeping all transactions
)
final.tail(12) #Displaying Results now including Transaction count and Total revenue
```

```python
#Creating pivot table for New York

ny_data = revenue[revenue['State'] == 'New York'] #Filters the revenue to show data for only NY

ny_pivot = ny_data.pivot_table( #Creating table
    index='Month', #Index makes months rows
    columns='Category', #Creates separate columns for Categories
    values='Total_Revenue', #Fills revenue values 
    fill_value=0 #Replaces any missing value
)

#Creating line plot from pivot 

ax=ny_pivot.plot(figsize=(12,6),linewidth=3 #Setting chart size and linewidth for view 

plt.title('Monthly Revenue by Category – New York') #Add a title to the chart
plt.xlabel('Month') #Label the X-axis
plt.ylabel('Total Revenue') #Label the Y-axis
plt.xticks(rotation=45) #Rotates ticks for view

ax.yaxis.set_major_formatter(ticker.StrMethodFormatter('${x:,.0f}')) #Formatting Y-axis to display as currency

#Adding Legend for context

plt.legend(title='Category', bbox_to_anchor=(1.05, 1)) #Moves legend outside of chart
plt.tight_layout() #Adjust for no overlapping
plt.show() #Display chart results
```

```python
#Creating pivot table for Connecticut

ct_data = revenue[revenue['State'] == 'Connecticut'] #Filters the revenue to show data for only CT

ct_pivot = ct_data.pivot_table( #Creating table
    index='Month', #Index #Creates separate columns for Categories months rows#Creates seperate columns for Categories
    columns='Category', #Creates seperate columns for Categories
    values='Total_Revenue', #Fills revenue values 
    fill_value=0 #Replaces any missing value
)

#Creating line plot from pivot 

ax=ct_pivot.plot(figsize=(12,6),linewidth=3)  #Setting chart size and linewidth for view

plt.title('Monthly Revenue by Category – Connecticut')  #Setting chart size and linewidth for view
plt.xlabel('Month') #Label the X-axis
plt.ylabel('Total Revenue') #Label the Y-axis
plt.xticks(rotation=45) #Rotates ticks for view

ax.yaxis.set_major_formatter(ticker.StrMethodFormatter('${x:,.0f}')) #Formatting Y-axis to display as currency

plt.legend(title='Category', bbox_to_anchor=(1.05, 1)) #Moves legend outside of chart#Adjust for no overlapping
plt.tight_layout() #Adjust for no overlapping
plt.show() #Display chart results
```

```python
# Creating side by side chart overview 

fig, axes = plt.subplots(1, 2, figsize=(16,6),linewidth=5) # Creating figure and two subplots 

#NEW YORK PLOT 

ny_pivot.plot(ax=axes[0]) #Plot the New York pivot table/Places NY on left of of subplot 
axes[0].set_title('New York') #Labels the first chart as New York
axes[0].yaxis.set_major_formatter(ticker.StrMethodFormatter('${x:,.0f}')) #Format the y-axis values for New York as currency
# ${x:,.0f} → adds dollar sign, commas, and removes decimal places

#CONNECTICUT plot

ct_pivot.plot(ax=axes[1]) #Plot the Connecticut pivot table/Places CT on right of subplot
axes[1].set_title('Connecticut') #Labels the second chart as Connecticut 
axes[1].yaxis.set_major_formatter(ticker.StrMethodFormatter('${x:,.0f}')) #Format the y-axis values Connecticut as currency
# ${x:,.0f} → adds dollar sign, commas, and removes decimal places

plt.tight_layout() #Adjust spacing between subplots automatically
plt.show() #Display the final result with both subplots
```
